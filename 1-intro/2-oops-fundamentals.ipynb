classes 

```python
class Animal:
    species = "dog"
    def __init__(self,name,color):
        self.name = name
        self.color = color 
    
    def description(self):
        return f"{self.name} has color {self.color}"

    @property 
    def getcolor(self):
        return f"{self.color}"

    @classmethod 
    def speak(cls):
        return f"{cls.species} says hello"

    @staticmethod
    def run():
        return f"animal is running" 

animal = Animal(name="dog",color="black")
print(animal.description())
print(animal.color)
print(animal.getcolor)
print(Animal.speak())
print(Animal.run())
```

Inheritance (IS A) relationship , Pet is an Animal 


```python
class Animal:

    def __init__(self,name,color):
        self.name = name 
        self.color = color 

    def print_info(self):
        print(f"this animal has name {self.name} & color {self.color}")


class Pet(Animal):

    def __init__(self,name,color,breed,sex):
        super().__init__(name,color)
        self.breed = breed 
        self.sex = sex

    @property
    def get_sex(self):
        print(f"{self.sex}")

pet = Pet("cat","black","persian","male")

pet.print_info()
pet.get_sex
```

Composition (HAS A) relationship , Car has a Engine 

```python
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        return "Vroom! Engine is running."

class Car:
    def __init__(self, model, horsepower):
        self.model = model
        # 🧩 Composition: We put an Engine object INSIDE the Car object
        self.engine = Engine(horsepower) 

    def start_car(self):
        # The Car delegates the starting work to its Engine
        return f"{self.model} says: {self.engine.start()}"

# Let's create a car
my_car = Car("Mustang", 450)

# Start the car
print(my_car.start_car()) 
```

Decorators

```python
#It takes an existing function, wraps it inside some extra code, and changes or extends 
# its behavior without you having to modify the original function's code.

def my_decorator(func):
    def wrapper():
        print("[before] executing fn")
        func()
        print("[after] executing fn")
    return wrapper

# open_gift =  my_decorator(open_gift)
@my_decorator 
def open_gift():
    print("this is a toy")

open_gift()

```

```python
for item in [1,2,3,4,5]:
    print(item)

print("\n")
print("\n")
#genrators (will output asynchronously)
import time 
def my_func():
    yield 1
    time.sleep(3)
    yield 2
    time.sleep(3)

values = my_func()
print(next(values))
print(next(values))
print("this is value ", *my_func())
```

*args & **kwargs (kewords arguments)

```python
def print_nums(*args):
    print(type(args))
    print(args)
    print("\n")

print_nums(1,2,3,4,5)

def print_profile(**kwargs):
    print(type(kwargs))
    print(kwargs)
    print("\n")


print_profile(name="anuj",age=27)


## also use to unpack values 
nums = [10,20,30]
print_nums(*nums)

profile = {
    "name" : "ayush",
    "age" : 27
}

print_profile(**profile)

```
