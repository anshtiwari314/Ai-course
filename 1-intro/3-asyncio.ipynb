#### blocking & non-blocking asynchronous operations 

```python
import asyncio
import time 

async def task_A():
    print("[before] sleep A")
    await asyncio.sleep(2) 
    print("[after] sleep A")

async def task_B():
    print("[before] sleep B")
    await asyncio.sleep(1) 
    print("[after] sleep B")


## blocking synchronously main co-routine 

# async def main():
#     await task_A()
#     await task_B()
# start = time.perf_counter()



## non blocking asynchronously 
async def main():
    results = await asyncio.gather(
     task_A(),
     task_B()
    ) 
    print("results-->", results)

start = time.perf_counter()

await main()

print(f"time counter {time.perf_counter()-start:.2f}")
```

#### async generators & async for loop 

```python
import asyncio 

async def generate_numbers():

    for num in range(5):
        await asyncio.sleep(1)
        yield num 

async for i in generate_numbers():
    print(i)
```

Http calls

```python
import httpx 
import asyncio 

async def get_post():

    async with httpx.AsyncClient() as client:
       result = await client.get("https://jsonplaceholder.typicode.com/posts/1")
       return result.json()

data = await get_post()
print(data.keys())
#print(data)
```
