```python
from langchain_text_splitters import RecursiveCharacterTextSplitter,MarkdownHeaderTextSplitter , TokenTextSplitter
```

```python
wiki_google_content
with open('/content/wiki_google.txt', 'r') as f:
    wiki_google_content = f.read()

print(f"Content from wiki_google.txt (first 200 chars):\n{wiki_google_content[:200]}...")
```

```python
wiki_nvidia_content
with open('/content/wiki_nvidia.txt', 'r') as f:
    wiki_nvidia_content = f.read()

print(f"Content from wiki_nvidia.txt (first 200 chars):\n{wiki_nvidia_content[:200]}...")
```

```python
large_markdown_content = ''
with open('/content/large_markdown_dataset.md', 'r') as f:
    large_markdown_content = f.read()

print(f"Content from wiki_nvidia.txt (first 200 chars):\n{large_markdown_content[:200]}...")
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000, # here chunks_size = 1000 char
    chunk_overlap=200,
    separators=["\n\n", "\n", ".", "!", "?", ",", " ", ""],
    length_function=len,
)
```

```python
chunks = text_splitter.split_text(wiki_google_content)
```

```python
markdown_text_splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=[
        ("#", "Header 1"),
        ("##", "Header 2"),
        ("###", "Header 3"),
        ("####", "Header 4"),
    ]
)
```

```python
markdown_chunks = markdown_text_splitter.split_text(large_markdown_content)
```

```python
token_splitter= TokenTextSplitter(
    chunk_size=1000, #here chunks size = token size
    chunk_overlap=200
)
```

```python
token_chunks= token_splitter.split_text(large_markdown_content)
```

```python
for i in range(0,5):
  print(token_chunks[i],"\n\n\n\n")
```
